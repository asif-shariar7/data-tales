# Any animal except...


During the global exploration of the British empire from 1750 to 1835, many animals were brought back and exhibited to the English public for the first time. These "exotic" creatures were showcased and traded in precursors to zoos, known as menageries.

In this project we will investigate how British menageries affected the usage frequency for different animal words.

Explanation : Between 1750 and 1835, the British Empire explored many parts of the world. During these explorations, they brought unusual or unfamiliar animals back to Britain. These animals were then :<br>

Shown to the public<br>
Displayed in special places called menageries<br>
Sometimes traded or sold<br>

A menagerie was like an early version of a zoo.<br>

What is this project about?<br>

The project wants to find out : Did showing these exotic animals in British menageries make people use the names of those animals more often?

So, the main goal is to compare animal word usage before and after the animals became popular in menageries.


```python
import pandas as pd
import matplotlib.pyplot as plt

word_trends = pd.read_csv('animal-word-trends-menageries.csv')
word_trends
```

```python
def plot_menagerie_animal( animal ):
    plt.figure(figsize=(8, 3))
    trend = word_trends.query("word == @animal")
    plt.plot(trend['year'], trend['frequency'])
    plt.ylabel('Frequency per million')
    plt.title(f'The menagerie period and word frequency for "{animal}"')
    plt.axvspan(1750, 1835, color='magenta', alpha=0.2)
```

Note : This shows a relationship, but it does not prove that menageries caused the increase.

### Explanatory mechanism
We just observed two different trends for exotic animals:

- Exhibited animals showed an increase in word mentions during the menagerie period.
- Non-exhibited animals showed little change in word mentions during the same period.

A good data story conveys not just what happened, but also provides an explanation for why it might have happened.

Provide an explanatory mechanism for the differences in the trends for the two groups of exotic animals.

Explanation :<br>

Word usage increased during the menagerie period mainly for animals that people could actually see in menageries. Exhibited animals were shown to the public from 1750–1835, so repeated exposure may have made people use their names more often.
Animals that were difficult to keep alive or feed were rarely exhibited during this period. As a result, people had less exposure to them, so their word frequencies stayed mostly flat.
After 1835, better zoos and improved methods of keeping animals in captivity made some of these animals easier to see, and their word usage began to increase.
